## Preparación
Este archivo usa una copia didáctica del dataset Iris con **8 celdas vacías introducidas para practicar limpieza**. Son 150 flores, tres especies y cuatro mediciones. Los datos originales se guardan en `iris_original.csv` para comparar.

Coloca este notebook en la misma carpeta que los archivos CSV. Si aparece `FileNotFoundError`, revisa la carpeta abierta en VS Code.

```python
import pandas as pd

iris = pd.read_csv("iris_con_faltantes.csv")
print("Dataset cargado:", iris.shape)
display(iris.head(3))
```

**Comprobación esperada:** `(150, 5)`.

**Referencia de columnas:** `sepal_length_cm`, `sepal_width_cm`, `petal_length_cm`, `petal_width_cm`, `species`. Las medidas están expresadas en centímetros.

## Ejercicio 1. Explorar el dataset

a) Muestra las primeras cinco filas. b) Identifica las dimensiones. c) Consulta los tipos de columna y sus valores no nulos. d) Encuentra cuáles especies aparecen en la columna `species`.

**Pista:** Recuerda `head`, `shape`, `info` y `unique`. Debes encontrar **3 especies**.


```python
print(iris.head(5))
print()
print(iris.shape)
iris.info()
print()
print(iris['species'].unique())
```

## Ejercicio 2. Seleccionar columnas y posiciones

a) Crea una tabla que contenga solo `species` y `petal_length_cm`. b) Muestra **las cinco primeras filas y las dos primeras columnas** mediante posiciones, no nombres.

**Pista:** Usa una lista de columnas para (a) y `iloc` para (b). El final de un slice no se incluye.


```python
iris_sp= iris[['species' , 'petal_length_cm']]
print(iris_sp)

iris_cd= iris.iloc[:5, :2]
print(iris_cd)
```

## Ejercicio 3. Filtrar con una o dos condiciones

a) Selecciona solamente flores de la especie `setosa`. b) Selecciona flores de `virginica` cuyo `petal_length_cm` sea **mayor que 5.5**. c) Muestra cuántas filas tiene cada resultado.

**Pista:** Utiliza una máscara de comparación y `&`; cada condición debe ir entre paréntesis.


```python
iris_a= iris[iris['species'] == 'setosa']
print(iris_a)

print()

iris_b= iris[iris['species'] == 'virginica']
iris_b= iris_b[iris_b['petal_length_cm'] > 5.5]
print(iris_b)

print()

print("Cantidad de filas para la pregunta a:", iris_a.shape[0])
print("Cantidad de filas para la pregunta b:", iris_b.shape[0])
```

## Ejercicio 4. Contar y agrupar por especie

a) Cuenta cuántas flores hay de cada especie. b) Calcula el **promedio de `petal_length_cm` para cada especie**. c) Identifica cuál tiene el promedio mayor.

**Pista:** Busca `value_counts`, `groupby` y `mean`. La columna `petal_length_cm` no contiene faltantes en esta copia.


```python
conteo= iris['species'].value_counts()
print(conteo)
print()
Promedio= iris.groupby('species').mean()
print(Promedio)
print()
promedio_mayor= Promedio['petal_length_cm'].idxmax()
valor_promedio= Promedio['petal_length_cm'].max()
print("La especie con el mayor promedio de longitud de pétalo es:", promedio_mayor, "con un valor de:", valor_promedio)
```

## Ejercicio 5. Limpiar valores faltantes y quitar columnas

a) Cuenta cuántos valores faltan por columna. b) Crea **una copia** sin las filas que tengan valores vacíos y cuenta cuántas quedan. c) Crea otra copia donde rellenes únicamente los faltantes de `sepal_width_cm` con su mediana. d) Crea una tercera copia sin la columna `petal_width_cm`. **No modifiques `iris` directamente.**

**Pista:** Recuerda `isna().sum()`, `dropna()`, `fillna()`, `median()` y `drop(columns=[...])`. Esperamos **142 filas completas** antes de rellenar.


```python
Valores_faltantes= iris.isna().sum()
print("Valores faltantes por columna:")
print(Valores_faltantes)
print()
Copia_sin_filas_vacias= iris.dropna()
print("Cantidad de filas después de eliminar filas con valores faltantes:", Copia_sin_filas_vacias.shape[0])
print()
relleno_faltantes= iris.fillna(iris['sepal_width_cm'].mean())
print("Valores faltantes después del relleno:")
print(relleno_faltantes.isna().sum()) 
print()
Copia_sin_columna= iris.drop(columns=['petal_width_cm'])
print(Copia_sin_columna)

```

## Ejercicio 6. Transformar columnas

a) Crea en una copia una nueva columna `petal_length_mm` convirtiendo centímetros a milímetros. b) Crea otra columna `species_codigo` asignando `S`, `V` y `G` a las tres especies. c) Comprueba cinco filas.

**Pista:** Para la conversión basta una operación sobre toda la columna; para códigos puedes usar `.map()` con un diccionario. Si quieres, prueba también `.apply(lambda ...)`.


```python
iris_copia= iris.copy()
iris_copia['petal_length_mm']= iris_copia['petal_length_cm'] * 10
print()
codigos= {'setosa': 'S', 'versicolor': 'V', 'virginica': 'G'}
iris_copia['species_code']= iris_copia['species'].map(codigos)
print()
iris_copia.head(5)

```

## Ejercicio 7. Exportar el resultado

a) Guarda el DataFrame transformado del ejercicio 6 en `iris_procesado.csv`, **sin guardar la columna de índice**. b) Vuelve a leer el CSV con Pandas y comprueba que conserva 150 filas y las nuevas columnas.

**Pista:** Usa `to_csv` con `index=False` y después `read_csv`.


```python
iris_copia.to_csv('iris_procesado.csv', index=False)
print()
iris_leido= pd.read_csv('iris_procesado.csv')
print()
print("El número de filas y columnas del nuevo archivo es:", iris_leido.shape)
print("La lista de columnas es de:", iris_leido.columns.tolist())
```

## Ejercicio 8. Unir filas con `concat`

a) Construye dos DataFrames: primeras 50 filas y siguientes 50 filas de `iris`. b) Únelos verticalmente y comprueba el número total de filas.

**Pista:** Combina `iloc` y `pd.concat`; prueba `ignore_index=True`.


```python
iris_1= iris.iloc[0:50]
iris_2= iris.iloc[50:100]
print()
iris_concat= pd.concat([iris_1,iris_2], ignore_index=True)
print()
print("El número total de filas es:", len(iris_concat))
```

## Ejercicio 9. Relacionar tablas con `merge`
Crea una tabla pequeña con las columnas `species` y `codigo`: setosa→S, versicolor→V y virginica→G. Une esta tabla con `iris` mediante la especie. Verifica que aparecen 150 filas y una columna adicional. Los códigos son solamente etiquetas didácticas.

**Pista:** Puedes construir la tabla auxiliar con `pd.DataFrame({...})` y hacer un `merge` por `species`.


```python
tabla_codigos= pd.DataFrame({'species': ['setosa', 'versicolor', 'virginica'], 'species_code': ['S', 'V', 'G']})
print()
iris_modificado= pd.merge(iris, tabla_codigos, on='species', how='left')
print()
print("El número de filas es de:", len(iris_modificado))
print(iris_modificado.head(5))
```
