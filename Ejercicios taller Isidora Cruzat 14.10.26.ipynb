```python
Sospechosas = ["bloqueada","ultimo aviso","urgente","reembolso","urgente","ganaste","premio"]
Contador = 0
Mensaje = str(input("Escriba su mensaje:"))
for palabra in Sospechosas:
    if palabra in Mensaje:
        Contador += 1
print("El mensaje contiene", Contador, "palabras sospechosas")
        
```

```python
positivas = ["gracias","feliz","excelente","bueno","positivo"]
negativas = ["malo","horrible","pesimo","negativo","triste"]
Texto = str(input("Escriba su mensaje:"))
Contador_positivo = 0
Contador_negativo = 0
for palabra in positivas:
    if palabra in Texto:
        Contador_positivo += 1
for palabra in negativas:
    if palabra in Texto:
        Contador_negativo += 1
if Contador_positivo > Contador_negativo:
    print("El mensaje es positivo")
elif Contador_negativo > Contador_positivo:
    print("El mensaje es negativo")
else:
    print("El mensaje es neutral")
```

```python
salir = False
mensajes = str(input("Escriba su mensaje:"))
while salir == False:
    if "salir" in mensajes:
        salir = True
        print("Saliendo del programa...")
    else:
        mensajes = str(input("Escriba su mensaje:"))
```
